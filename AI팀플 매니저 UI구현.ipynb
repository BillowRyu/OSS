```python

import gradio as gr
from html import escape

# ==========================================
# TEAMFLOW | AI 팀플 매니저
# UI MVP - 수정 버전
# ==========================================

PROJECT_NAME = "TeamFlow AI 협업 플랫폼 런칭"

# 팀원 및 실제 역할
members = {
    "김병찬": "발표 및 아이디어 제공",
    "김민수": "LLM 코드 작성",
    "이창주": "Notion / Asana UI 자료 수집",
    "김의연": "Notion / Asana UI 자료 수집",
}

# 업무 상태
STATUS_OPTIONS = ["진행 중", "대기", "완료"]

# 상태별 색상
STATUS_COLORS = {
    "진행 중": "#F4C542",  # 노란색
    "대기": "#EF5350",     # 빨간색
    "완료": "#2EBD85",     # 초록색
}

# 업무 데이터
tasks = [
    {
        "id": "01",
        "name": "Notion UI 자료 조사",
        "description": "Notion의 화면 구성과 프로젝트 관리 UI 참고",
        "member": "이창주",
        "status": "진행 중",
        "date": "2026-11-09",
    },
    {
        "id": "02",
        "name": "Asana UI 자료 조사",
        "description": "Asana의 업무 목록과 칸반 UI 참고",
        "member": "김의연",
        "status": "진행 중",
        "date": "2026-11-09",
    },
    {
        "id": "03",
        "name": "UI 구성 및 아이디어 정리",
        "description": "팀 프로젝트 화면 구성과 발표 아이디어 제안",
        "member": "김병찬",
        "status": "진행 중",
        "date": "2026-11-11",
    },
    {
        "id": "04",
        "name": "LLM 코드 작성",
        "description": "AI 팀플 매니저의 LLM 코드 및 핵심 기능 작성",
        "member": "김민수",
        "status": "진행 중",
        "date": "2026-11-12",
    },
    {
        "id": "05",
        "name": "업무 관리 UI 구현",
        "description": "업무 목록과 담당자 및 상태 표시 UI 제작",
        "member": "김병찬",
        "status": "대기",
        "date": "2026-11-15",
    },
    {
        "id": "06",
        "name": "팀원별 업무 화면 구성",
        "description": "팀원별 담당 업무를 확인하는 화면 제작",
        "member": "김병찬",
        "status": "대기",
        "date": "2026-11-17",
    },
    {
        "id": "07",
        "name": "UI 및 LLM 코드 통합",
        "description": "UI와 기존 LLM 코드의 연결 및 테스트",
        "member": "김민수",
        "status": "대기",
        "date": "2026-11-20",
    },
    {
        "id": "08",
        "name": "발표 자료 및 최종 시연 준비",
        "description": "프로젝트 결과 정리 및 발표 준비",
        "member": "김병찬",
        "status": "대기",
        "date": "2026-11-26",
    },
]


# ==========================================
# 디자인 CSS
# ==========================================

css = """
body {
    background: #0b1220 !important;
    color: #e8edf7 !important;
}

.gradio-container {
    max-width: 1500px !important;
    background: #0b1220 !important;
}

.block, .panel {
    background: #172337 !important;
    border: 1px solid #2b3b52 !important;
    border-radius: 12px !important;
}

h1, h2, h3 {
    color: #f4f7fc !important;
}

label, p, span {
    color: #dce5f3;
}

button {
    border-radius: 8px !important;
}

button.primary {
    background: #6555e8 !important;
    color: white !important;
}

.tab-nav button {
    background: #172337 !important;
    color: #aabbd2 !important;
}

.tab-nav button.selected {
    background: #293c5d !important;
    color: white !important;
    border-bottom: 2px solid #8275ff !important;
}
"""


# ==========================================
# 업무 상태별 HTML 카드 생성
# ==========================================

def task_card(task):
    status = task["status"]
    color = STATUS_COLORS[status]

    return f"""
    <div style="
        background:#111c2e;
        border:1px solid #30415a;
        border-radius:12px;
        padding:16px;
        margin-bottom:10px;
        min-height:145px;
        box-sizing:border-box;
    ">
        <div style="
            display:flex;
            justify-content:space-between;
            align-items:center;
            gap:8px;
            margin-bottom:12px;
        ">
            <span style="
                color:#8ea4c2;
                font-size:11px;
                font-weight:700;
            ">TASK #{escape(task['id'])}</span>

            <span style="
                background:{color}20;
                color:{color};
                border:1px solid {color}80;
                border-radius:6px;
                padding:4px 8px;
                font-size:11px;
                font-weight:700;
                white-space:nowrap;
            ">{escape(status)}</span>
        </div>

        <div style="
            color:#ffffff;
            font-size:14px;
            font-weight:700;
            line-height:1.5;
            margin-bottom:8px;
        ">{escape(task['name'])}</div>

        <div style="
            color:#9fb0c8;
            font-size:12px;
            line-height:1.6;
            margin-bottom:14px;
        ">{escape(task['description'])}</div>

        <div style="
            display:flex;
            justify-content:space-between;
            align-items:center;
            gap:6px;
            border-top:1px solid #2b3b52;
            padding-top:10px;
        ">
            <span style="
                color:#c4d2e3;
                font-size:11px;
        ">👤 {escape(task['member'])}</span>

            <span style="
                color:#8ea4c2;
                font-size:10px;
        ">📅 {escape(task['date'])}</span>
        </div>
    </div>
    """


# ==========================================
# 가로형 업무 보드
# ==========================================

def render_board():
    columns = {
        "진행 중": [],
        "대기": [],
        "완료": [],
    }

    for task in tasks:
        columns[task["status"]].append(task)

    html = """
    <div style="
        display:grid;
        grid-template-columns:repeat(3,minmax(250px,1fr));
        gap:14px;
        align-items:start;
        width:100%;
    ">
    """

    for status, items in columns.items():
        color = STATUS_COLORS[status]

        html += f"""
        <div style="
            background:#172337;
            border:1px solid #2b3b52;
            border-radius:12px;
            padding:12px;
            min-width:0;
        ">
            <div style="
                display:flex;
                justify-content:space-between;
                align-items:center;
                border-bottom:1px solid #30415a;
                padding:5px 2px 12px;
                margin-bottom:12px;
            ">
                <div style="
                    display:flex;
                    align-items:center;
                    gap:8px;
                ">
                    <div style="
                        width:9px;
                        height:9px;
                        background:{color};
                        border-radius:50%;
                    "></div>

                    <span style="
                        color:#ffffff;
                        font-size:14px;
                        font-weight:700;
                    ">{status}</span>
                </div>

                <span style="
                    background:#26364d;
                    color:#c4d2e3;
                    border-radius:6px;
                    padding:3px 8px;
                    font-size:11px;
        ">{len(items)}</span>
            </div>
        """

        if items:
            for task in items:
                html += task_card(task)
        else:
            html += """
            <div style="
                padding:25px 10px;
                text-align:center;
                color:#71849e;
                font-size:12px;
                border:1px dashed #30415a;
                border-radius:10px;
            ">등록된 업무가 없습니다.</div>
            """

        html += "</div>"

    html += "</div>"

    return html


# ==========================================
# 업무 상태 변경
# ==========================================

def update_task(task_id, new_status):
    for task in tasks:
        if task["id"] == task_id:
            task["status"] = new_status
            break

    return render_board()


# ==========================================
# 팀원별 업무
# ==========================================

def filter_member(member):
    selected = [
        task for task in tasks
        if task["member"] == member
    ]

    if not selected:
        return """
        <div style="
            background:#172337;
            border-radius:12px;
            padding:25px;
            color:#a9bad2;
        ">담당 업무가 없습니다.</div>
        """

    html = """
    <div style="
        display:grid;
        grid-template-columns:repeat(2,minmax(0,1fr));
        gap:12px;
    ">
    """

    for task in selected:
        html += task_card(task)

    html += "</div>"

    return html


# ==========================================
# 팀원 역할 카드
# ==========================================

def render_members():
    html = """
    <div style="
        display:grid;
        grid-template-columns:repeat(2,minmax(0,1fr));
        gap:12px;
    ">
    """

    for name, role in members.items():
        count = len([
            task for task in tasks
            if task["member"] == name
        ])

        html += f"""
        <div style="
            background:#172337;
            border:1px solid #2b3b52;
            border-radius:12px;
            padding:16px;
        ">
            <div style="
                color:#ffffff;
                font-size:15px;
                font-weight:700;
                margin-bottom:8px;
            ">👤 {escape(name)}</div>

            <div style="
                color:#a9bad2;
                font-size:12px;
                line-height:1.7;
        ">{escape(role)}</div>

            <div style="
                color:#8275ff;
                font-size:11px;
                margin-top:12px;
        ">담당 업무 {count}개</div>
        </div>
        """

    html += "</div>"

    return html


# ==========================================
# 대시보드 정보
# ==========================================

def render_dashboard():
    total = len(tasks)
    completed = sum(
        task["status"] == "완료"
        for task in tasks
    )
    ongoing = sum(
        task["status"] == "진행 중"
        for task in tasks
    )
    waiting = sum(
        task["status"] == "대기"
        for task in tasks
    )

    progress = round(
        completed / total * 100
    ) if total else 0

    return f"""
    <div style="
        display:grid;
        grid-template-columns:repeat(4,minmax(0,1fr));
        gap:12px;
    ">
        <div style="
            background:#172337;
            border:1px solid #2b3b52;
            border-radius:12px;
            padding:18px;
        ">
            <div style="color:#91a5c0;font-size:12px;">전체 업무</div>
            <div style="color:#ffffff;font-size:30px;font-weight:800;margin-top:8px;">{total}</div>
        </div>

        <div style="
            background:#172337;
            border:1px solid #2b3b52;
            border-radius:12px;
            padding:18px;
        ">
            <div style="color:#F4C542;font-size:12px;">진행 중</div>
            <div style="color:#F4C542;font-size:30px;font-weight:800;margin-top:8px;">{ongoing}</div>
        </div>

        <div style="
            background:#172337;
            border:1px solid #2b3b52;
            border-radius:12px;
            padding:18px;
        ">
            <div style="color:#EF5350;font-size:12px;">대기</div>
            <div style="color:#EF5350;font-size:30px;font-weight:800;margin-top:8px;">{waiting}</div>
        </div>

        <div style="
            background:#172337;
            border:1px solid #2b3b52;
            border-radius:12px;
            padding:18px;
        ">
            <div style="color:#2EBD85;font-size:12px;">완료</div>
            <div style="color:#2EBD85;font-size:30px;font-weight:800;margin-top:8px;">{completed}</div>
        </div>
    </div>

    <div style="
        background:#172337;
        border:1px solid #2b3b52;
        border-radius:12px;
        padding:18px;
        margin-top:14px;
    ">
        <div style="
            display:flex;
            justify-content:space-between;
            align-items:center;
            margin-bottom:12px;
        ">
            <span style="color:#c4d2e3;font-size:13px;">프로젝트 완료율</span>
            <span style="color:#8275ff;font-size:16px;font-weight:700;">{progress}%</span>
        </div>

        <div style="background:#30415a;height:8px;border-radius:10px;">
            <div style="
                background:#8275ff;
                width:{progress}%;
                height:8px;
                border-radius:10px;
            "></div>
        </div>
    </div>
    """


# ==========================================
# 회의 분석 시연
# ==========================================

def analyze_meeting(text):
    if not text.strip():
        return "회의 내용을 먼저 입력해주세요."

    return """
    ### 회의 요약

    입력된 회의 내용을 바탕으로 분석 결과를 표시합니다.

    **주요 논의**
    - 프로젝트 진행 상황 확인
    - 담당 업무 및 일정 검토
    - 다음 작업 계획 정리

    **업무 초안**
    - 회의 내용을 바탕으로 업무 정리
    - 담당자 및 마감일 확인
    - 진행 상황 공유

    ※ 현재는 UI 시연용 예시 결과입니다.
    실제 AI 회의 분석 기능은 연결되지 않았습니다.
    """


# ==========================================
# 전체 UI
# ==========================================

with gr.Blocks(
    title="TEAMFLOW | AI 팀플 매니저",
    theme=gr.themes.Base(),
    css=css
) as app:

    # 헤더
    gr.HTML("""
    <div style="
        background:#111c2e;
        border:1px solid #2b3b52;
        border-radius:14px;
        padding:20px 24px;
        margin-bottom:16px;
    ">
        <div style="display:flex;align-items:center;gap:12px;">
            <div style="
                background:#6555e8;
                width:42px;height:42px;
                border-radius:12px;
                display:flex;
                align-items:center;
                justify-content:center;
                font-size:24px;
            ">⚡</div>

            <div>
                <div style="
                    font-size:23px;
                    font-weight:800;
                    color:#ffffff;
                ">TEAMFLOW</div>

                <div style="
                    font-size:11px;
                    letter-spacing:2px;
                    color:#8298b8;
                ">AI PROJECT MANAGER</div>
            </div>
        </div>
    </div>
    """)

    gr.Markdown(f"# {PROJECT_NAME}")

    with gr.Tabs():

        # ----------------------------------
        # 대시보드
        # ----------------------------------
        with gr.Tab("📊 대시보드"):

            gr.Markdown("### 프로젝트 현황")

            dashboard_html = gr.HTML(
                value=render_dashboard()
            )

            gr.Markdown("### 팀원 및 역할")

            members_html = gr.HTML(
                value=render_members()
            )

            gr.Markdown("### 업무 현황")

            dashboard_board = gr.HTML(
                value=render_board()
            )

        # ----------------------------------
        # 업무 관리
        # ----------------------------------
        with gr.Tab("📋 업무 관리"):

            gr.Markdown("## 업무 관리")

            gr.Markdown(
                "업무를 상태별로 나누어 확인하고 관리합니다."
            )

            board = gr.HTML(
                value=render_board()
            )

            gr.Markdown("### 업무 상태 변경")

            with gr.Row():

                task_select = gr.Dropdown(
                    choices=[
                        task["id"] + " | " + task["name"]
                        for task in tasks
                    ],
                    label="업무 선택"
                )

                status_select = gr.Dropdown(
                    choices=STATUS_OPTIONS,
                    value="진행 중",
                    label="변경할 상태"
                )

            update_button = gr.Button(
                "업무 상태 변경",
                variant="primary"
            )

            def change_status(selected_task, status):
                if not selected_task:
                    return render_board()

                task_id = selected_task.split(" | ")[0]

                return update_task(task_id, status)

            update_button.click(
                fn=change_status,
                inputs=[task_select, status_select],
                outputs=[board]
            )

        # ----------------------------------
        # 팀원별 업무
        # ----------------------------------
        with gr.Tab("👥 팀원별 업무"):

            gr.Markdown("## 팀원별 업무")

            member_select = gr.Dropdown(
                choices=list(members.keys()),
                value="김병찬",
                label="팀원 선택"
            )

            member_tasks = gr.HTML(
                value=filter_member("김병찬")
            )

            member_select.change(
                fn=filter_member,
                inputs=member_select,
                outputs=member_tasks
            )

            gr.Markdown("### 팀원 역할")

            member_roles = gr.HTML(
                value=render_members()
            )

        # ----------------------------------
        # 회의 분석
        # ----------------------------------
        with gr.Tab("📝 회의 분석"):

            gr.Markdown("## 회의 분석")

            meeting_input = gr.Textbox(
                label="회의 내용",
                placeholder="회의 내용을 입력하세요.",
                lines=8
            )

            analyze_button = gr.Button(
                "회의 분석",
                variant="primary"
            )

            meeting_output = gr.Markdown(
                "회의 내용을 입력하고 분석 버튼을 눌러주세요."
            )

            analyze_button.click(
                fn=analyze_meeting,
                inputs=meeting_input,
                outputs=meeting_output
            )


app.launch()
```
